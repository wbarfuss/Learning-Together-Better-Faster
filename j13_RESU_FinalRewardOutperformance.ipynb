# Result | Final rewards {#sec-results-final-rewards}


Here, we ask the question, how the level of reward outperformance at convergences depend on the environmental uncertainty and the learning model. To do so, we compare the final rewards obtained by agents in the single-agent and the two-agent uncertain decision problem across different levels of environmental uncertainty and both learning models (model-free vs. model-based). We define the reward outperformance as the difference between the two-agent reward $R_\text{two}$,and the single-agent reward $R_\text{single}$, normalized by the single-agent reward, $(R_\text{two} - R_\text{single}) / {R_\text{single}}.$ We start by implementing the necessary imports and configuration for plotting,

### Visualizing the reward outerpformance

Finally, we can visualize the reward outperformance for different levels of environmental uncertainty and both learning models.

```python
baseProbabilities = [0.5, 0.52, 0.55, 0.58,
                     0.6, 0.7, 0.9]  
NoiVar_MBA_list = [NoiVar_MBA_pA05, NoiVar_MBA_pA052, NoiVar_MBA_pA055, NoiVar_MBA_pA058,
                   NoiVar_MBA_pA06, NoiVar_MBA_pA07, NoiVar_MBA_pA09]
NoiVar_MFA_list = [NoiVar_MFA_pA5, NoiVar_MFA_pA52, NoiVar_MFA_pA55, NoiVar_MFA_pA058,
                   NoiVar_MFA_pA06, NoiVar_MFA_pA07, NoiVar_MFA_pA09]
```

```python
#| fig-cap: Final reward outperformance versus observational uncertainty for different state uncertainties and both learning dynamics
fsf = 0.55
fig, ax = plt.subplots(1, 1, figsize=(fsf*16, fsf*6))
cmap = plt.cm.plasma
norm = colors.Normalize(vmin=0.1, vmax=0.5)

ax.set_xlabel('Observational uncertainty $\\sigma$')
ax.set_ylabel('Reward outperformance\nat convergence');

for NoiVar_MBA, NoiVar_MFA, baseProbability in zip(NoiVar_MBA_list[::-1], 
                                                   NoiVar_MFA_list[::-1],
                                                   baseProbabilities[::-1]):
    col = cmap(norm(1-baseProbability))
    CInk_MBA = collective_intelligence(NoiVar_MBA.ta_Rnki.mean(-1),
                                   NoiVar_MBA.sa_Rnki.mean(-1))
    ax.plot(noise_levels, CInk_MBA.mean(-1), marker="+", ms=7, lw=0.5, color=col, 
            label=f'base probability: {baseProbability}')
    
    CInk_MFA = collective_intelligence(NoiVar_MFA.ta_Rnki.mean(-1),
                                   NoiVar_MFA.sa_Rnki.mean(-1))
    ax.plot(noise_levels, CInk_MFA.mean(-1), marker="x", ms=7, lw=0.5, color=col, 
            label=f'base probability: {baseProbability}')

# Add colorbar
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
cbar = plt.colorbar(sm, ax=ax, pad=0.025)
cbar.set_label('State uncertainty $p_B=1-p_A$')
cbar.ax.set_ylim(0.098, 0.502)

for bP in baseProbabilities:
    cbar.ax.plot([0.1, 0.8], [1-bP, 1-bP], color='springgreen', lw=3)

# Add legend to showcase MBA and MFA markers
prms = dict(ls="", color='k', ms=8)
legend_elements = [
        Line2D([0], [0], marker='+', **prms, label='Model-based agents'),
        Line2D([0], [0], marker='x', **prms, label='Model-free agents')]
ax.legend(handles=legend_elements, loc='lower right');

plt.tight_layout()
plt.subplots_adjust(right=1.05)
plt.savefig("./_figs/FinalRewardOutperformance.png", dpi=300)
```
