```python
 #| export
ddirs = dict(lowlevel_ddir=lowlevel_datadictionary,
             highlevel_ddir=highlevel_datadictionary)

mobaparams = dict(agentclass=LearningAgents, maxT=5000, number_of_initial_policies_k=25)
mofeparams = dict(agentclass=ModelFreeAgents, maxT=250000, number_of_initial_policies_k=10)
```

### Obtain data

Then, we create a small summary function to obtain the simulation results for different noise levels and state uncertainties.

```python
 #| export
def obtain_all_noise_variation_comparisons():
    NoiVar_MBA_pAhigh = obtain_noise_variation_comparison(
        probabilityA=0.5, 
        noise_levels=noise_levels,
        **mobaparams, **ddirs)

    NoiVar_MFA_pAhigh = obtain_noise_variation_comparison(
        probabilityA=0.5, 
        noise_levels=noise_levels,
        **mofeparams, **ddirs)

    NoiVar_MBA_pAmid = obtain_noise_variation_comparison(
        probabilityA=0.58, 
        noise_levels=noise_levels,
        **mobaparams, **ddirs)

    NoiVar_MFA_pAmid = obtain_noise_variation_comparison(
        probabilityA=0.58, 
        noise_levels=noise_levels,
        **mofeparams, **ddirs)

    NoiVar_MBA_pAlow = obtain_noise_variation_comparison(
        probabilityA=0.7, 
        noise_levels=noise_levels,
        **mobaparams, **ddirs)

    NoiVar_MFA_pAlow = obtain_noise_variation_comparison(
        probabilityA=0.7, 
        noise_levels=noise_levels,
        **mofeparams, **ddirs)
    
    return (NoiVar_MBA_pAhigh, NoiVar_MFA_pAhigh,
            NoiVar_MBA_pAmid, NoiVar_MFA_pAmid,
            NoiVar_MBA_pAlow, NoiVar_MFA_pAlow)
```
